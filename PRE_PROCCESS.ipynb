```python
# ==========================================
# 6. تعدیل تمام ستون‌های قیمتی
# ==========================================

import pandas as pd
import numpy as np


def adjust_price_history(price_df, events):

    # ------------------------------------------
    # کپی از داده‌ها
    # ------------------------------------------
    df = price_df.copy()
    events = events.copy()

    # ------------------------------------------
    # اطمینان از datetime بودن تاریخ‌ها
    # ------------------------------------------
    if "DATE" in df.columns:
        df["DATE"] = pd.to_datetime(df["DATE"])
        df = df.set_index("DATE")

    df.index = pd.to_datetime(df.index)
    events["date"] = pd.to_datetime(events["date"])

    # مرتب‌سازی
    df = df.sort_index()
    events = events.sort_values("date", ascending=False)

    # ------------------------------------------
    # ستون‌های قیمتی
    # ------------------------------------------
    price_columns = [
        "FIRST",
        "HIGH",
        "LOW",
        "CLOSE",
        "OPEN",
        "LAST"
    ]

    # ------------------------------------------
    # ساخت ستون‌های تعدیل‌شده
    # ------------------------------------------
    for col in price_columns:
        df[f"adj_{col.lower()}"] = df[col].astype(float)

    # ارزش معاملات
    df["adj_value"] = df["VALUE"].astype(float)

    # ------------------------------------------
    # اعمال رویدادها
    # از جدیدترین به قدیمی‌ترین
    # ------------------------------------------
    for _, event in events.iterrows():

        event_date = event["date"]
        event_type = event["type"]
        event_value = float(event["value"])

        # داده‌های قبل از رویداد
        mask = df.index < event_date

        # اگر داده‌ای قبل از رویداد وجود ندارد
        if not mask.any():
            continue

        # ======================================
        # افزایش سرمایه
        # ======================================
        if event_type == "capital_bonus":

            factor = 1 + event_value

            if factor > 0:

                for col in price_columns:
                    df.loc[
                        mask,
                        f"adj_{col.lower()}"
                    ] /= factor

                df.loc[mask, "adj_value"] /= factor

        # ======================================
        # سود نقدی
        # ======================================
        elif event_type == "dividend":

            dividend = event_value

            # آخرین قیمت خام قبل از رویداد
            before_event = df.loc[
                mask,
                "CLOSE"
            ]

            if len(before_event) == 0:
                continue

            P = float(before_event.iloc[-1])

            # جلوگیری از خطا
            if P <= 0:
                continue

            factor = (P - dividend) / P

            # فقط اگر ضریب معتبر باشد
            if factor > 0:

                for col in price_columns:

                    df.loc[
                        mask,
                        f"adj_{col.lower()}"
                    ] *= factor

                df.loc[
                    mask,
                    "adj_value"
                ] *= factor

    return df


# ==========================================
# آماده‌سازی داده اصلی
# ==========================================

price_df = df.copy()

# اگر DATE هنوز ستون است، تبدیلش می‌کنیم
if "DATE" in price_df.columns:

    price_df["DATE"] = pd.to_datetime(
        price_df["DATE"]
    )

    price_df = price_df.set_index("DATE")

else:

    price_df.index = pd.to_datetime(
        price_df.index
    )


price_df = price_df.sort_index()


# ==========================================
# اجرای تعدیل
# ==========================================

adjusted_df = adjust_price_history(
    price_df,
    events
)


# ==========================================
# بررسی نتیجه
# ==========================================

print("تعدیل با موفقیت انجام شد.")

print("\nستون‌های تعدیل‌شده:")
print([
    col for col in adjusted_df.columns
    if col.startswith("adj_")
])

print("\nنمونه داده:")
print(
    adjusted_df[
        ["CLOSE", "adj_close"]
    ].head(10)
)
```

```python
# ==========================================
# نمایش قیمت خام و تعدیل‌شده به تفکیک هر سال
# ==========================================

import matplotlib.pyplot as plt
import pandas as pd

# کپی از adjusted_df
plot_df = adjusted_df.copy()

# ------------------------------------------
# آماده‌سازی تاریخ
# ------------------------------------------
if "DATE" in plot_df.columns:
    plot_df["DATE"] = pd.to_datetime(plot_df["DATE"])
    plot_df = plot_df.sort_values("DATE")
    plot_df = plot_df.set_index("DATE")
else:
    plot_df.index = pd.to_datetime(plot_df.index)
    plot_df = plot_df.sort_index()

# ------------------------------------------
# سال‌های موجود در داده
# ------------------------------------------
years = sorted(plot_df.index.year.unique())

print("سال‌های موجود:")
print(years)

# ------------------------------------------
# رسم هر سال به صورت جداگانه
# ------------------------------------------
for year in years:

    year_df = plot_df[
        plot_df.index.year == year
    ]

    if len(year_df) == 0:
        continue

    plt.figure(figsize=(16, 6))

    # قیمت خام
    if "CLOSE" in year_df.columns:
        plt.plot(
            year_df.index,
            year_df["CLOSE"],
            label="Original CLOSE",
            linewidth=1,
            alpha=0.6
        )

    # قیمت تعدیل‌شده
    if "adj_close" in year_df.columns:
        plt.plot(
            year_df.index,
            year_df["adj_close"],
            label="Adjusted CLOSE",
            linewidth=1.5
        )

    plt.title(
        f"فولاد - مقایسه قیمت خام و تعدیل‌شده - سال {year}"
    )

    plt.xlabel("Date")
    plt.ylabel("Price")

    plt.legend()
    plt.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()
```
